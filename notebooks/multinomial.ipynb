```python
from scipy.stats import multinomial
import math
```

## PMF
$$
\begin{align*}
&P(X_{1}=k+{1},..., X_{m}=k_{m}) = \frac{n!}{\prod_{i=1}^{m}k_{i}!}\prod_{i=1}^{m}p_{i}^{k_{i}}
\end{align*}
$$

```python
k1, k2, k3 = 5, 3, 2
n = 10
# p must be summed up to 1
p = [.5,.3,.2]
dist = multinomial(n, p)
```

```python
dist.pmf([k1, k2, k3])
```

```python
combination = math.factorial(n)/math.prod([math.factorial(i) for i in [k1, k2, k3]])
combination
```

```python
prob = math.prod([0.5**5, 0.3**3, 0.2**2])
prob
```

```python
combination*prob
```
