```python
from scipy.stats import expon
import numpy as np
import math
```

```python
# server fail 0.5 times/year
# what is the chance of fail in the first 6 month
_lambda = 0.5
year = 12
month = 6
x = month/year
dist = expon(scale=1/_lambda)
dist.pdf(x), dist.cdf(x), dist.sf(x)
```

```python
1-np.exp(-_lambda*x)
```
