```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

n = 10000
samples = dist.rvs(size=n, random_state=555)

sns.displot(samples, kde=True, stat="density")

x = np.linspace(samples.min(), samples.max(), 300)
plt.plot(x, dist.pdf(x), color='red', label="theoretical PDF")
plt.legend()
```

## PMF
$$
\begin{align*}
&\text{k} = \text{success} \\
&\text{n} = \text{trials} \\
&\text{p} = \text{probability} \\

&P(X=k)=\binom{n}{k}p^{k}(1-p)^{n-k}
\end{align*}
$$

## Expected Value (mean)

\begin{align*}
