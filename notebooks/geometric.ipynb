```python
from scipy.stats import geom

p = 0.3
dist = geom(p)
```

```python
dist.pmf(3)
```

```python
dist.cdf(3)
```

```python
dist.mean()
```

```python
dist.var()
```

```python
dist.std()
```

```python
dist.rvs(size=10, random_state=555)
```

```python
import seaborn as sns

n = 10000
samples = dist.rvs(size=n, random_state=555)

sns.displot(samples,)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

n = 10000
samples = dist.rvs(size=n, random_state=555)

sns.displot(samples, kde=True, stat="density")

x = np.linspace(samples.min(), samples.max(), 300)
plt.plot(x, dist.pdf(x), color='red', label="theoretical PDF")
plt.legend()
```
