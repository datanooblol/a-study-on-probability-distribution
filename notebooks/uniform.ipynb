```python
from scipy.stats import uniform
import numpy as np

a, b = 0, 10
dist = uniform(loc=a, scale=b - a)   # scipy's uniform is parameterized by loc/scale, not a/b directly
```

```python
[(i, dist.pdf(i)) for i in range(1, b+1, 1)]
```

```python
[(i, dist.cdf(i)) for i in range(1, b+1, 1)]
```

```python
[(i, dist.ppf(i)) for i in np.arange(0.0, 1.1, 0.1)]
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

n = 10000
samples = dist.rvs(size=n, random_state=555)

sns.displot(samples, kde=True, stat="density")

x = np.linspace(samples.min(), samples.max(), 300)
plt.plot(x, dist.pdf(x), color='red', label="theoretical PDF")
plt.legend()
```
