```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

n = 10000
samples = dist.rvs(size=n, random_state=555)

sns.displot(samples, kde=True, stat="density")

x = np.linspace(samples.min(), samples.max(), 300)
plt.plot(x, dist.pdf(x), color='red', label="theoretical PDF")
plt.legend()
```

## PMF
$$ 
P(X=k)=p^{k}(1-p)^{1-k}, \ for \ k \in \{0,1\}
$$

```python
# p = success
p = 0.3
dist = bernoulli(p)
```

```python
[dist.pmf(k).round(1) for k in [0,1]]
```

```python
n = 100
dist.rvs(size=n, random_state=555)
```
