```python
from scipy.stats import lognorm, norm
import numpy as np
```

## PDF
$$
f(x)=\frac{1}{x\sigma\sqrt{2\pi}}e^{\frac{(ln \ x-\mu)^{2}}{2\sigma^{2}}}
$$

```python
mu, sigma = 11.2, 0.75
dist = lognorm(s=sigma, scale=np.exp(mu))
```

```python
k = 100_000
np.log(k)
```

```python
dist.sf(k), 1-dist.cdf(k)
```

```python
z_score = (np.log(k)-mu)/sigma
z_score
```

```python
z_score
```

```python
# change log world into normal world
_norm = norm(loc=0, scale=1)
```

```python
_norm.sf(z_score), 1-_norm.cdf(z_score)
```

```python
import numpy as np
from scipy.stats import lognorm, norm

mu, sigma = 11.2, 0.75
x = 100_000

# Method 1: manual — log the value, convert to z-score, use standard normal
z = (np.log(x) - mu) / sigma
p_manual = 1 - norm.cdf(z)

# Method 2: direct — let lognorm handle the transformation internally
dist = lognorm(s=sigma, scale=np.exp(mu))
# .sf = 1-.cdf
p_direct = dist.sf(x)

print("manual: ", p_manual)
print("direct: ", p_direct)
assert np.isclose(p_manual, p_direct), "mismatch between manual and direct methods"
print("match ✓")
```
