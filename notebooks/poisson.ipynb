```python
from scipy.stats import poisson
import numpy as np
import math
```

## PMF

$$
P(X=k)=\frac{\lambda^{k}e^{-\lambda}}{k!}
$$

```python
# if the case of 5 call per hour, then rate = 1
# if the case of 5 call per half an hour, then rate = 0.5
rate = 1
# number of observed successes
obs = 10
# expected value of observed event with correct rate
_lambda = obs*rate
# number of expected event to occur during _lambda
k = 3
dist = poisson(_lambda)
```

```python
((_lambda**k)*np.exp(-_lambda))/math.factorial(k)
```

```python
# pmf=exact event, cdf=event and below
dist.pmf(k), dist.cdf(k)
```

```python
dist.mean(), dist.std()
```

```python
# In 8 hours, sales-rep can sell 10 shirts in average
# What is the probability that he sells 3 shirts in 8 hours?

rate = 1
# number of observed successes
obs = 10
# expected value of observed event with correct rate
_lambda = obs*rate
# number of expected event to occur during _lambda
k = 3
dist = poisson(_lambda)
dist.pmf(k), dist.cdf(k)
```

```python
# In 8 hours, sales-rep can sell 10 shirts in average
# What is the probability that he sells 3 shirts in 4 hours?

rate = 0.5
# number of observed successes
obs = 10
# expected value of observed event with correct rate
_lambda = obs*rate
# number of expected event to occur during _lambda
k = 3
dist = poisson(_lambda)
dist.pmf(k), poisson.pmf(k, _lambda), dist.cdf(k), poisson.cdf(k, _lambda)
```

```python
dist.mean(), dist.var(), dist.std()
```
